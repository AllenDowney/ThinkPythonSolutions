```python
def compute(word1, word2):
    t = []
    for letter in word1:
        if letter in word2:
            t.append(letter)
    return ''.join(t)
```

```python
compute('abc', 'bcd')
```

```python
def lower_upper(word):
    first = word[0]
    rest = word[1:]
    word = first.lower() + rest.upper()
```

```python
class Day:
    
    def __init__(self, day):
        self.day = day

    def __str__(date):
        return f'Today is {date.day}'
    
print(Day('Wednesday'))
```

```python
def compute(d1, d2):
    res = {}
    for key in d1:
        res[key] = d1[key]
    for key in d2:
        res[key] = d2[key]
    return res
```

```python
d1 = dict(zip('abc', '123'))
d2 = dict(zip('cdf', '456'))
compute(d1, d2)
```
